```python
import numpy as np
```

```python
x= np.array([2,3,1])
weights = np.array([0.5,0.2,-0.4])
bias = 0.1
```

```python
#weighted sum
z=np.dot(x,weights)+bias
```

```python
#Activation function
y= np.maximum(0,z)
print("weighted sum:", z)
print("output:", y)
```

**Using different activation functions**

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))

def tanh(z):
  return np.tanh(z)

print("output-s:", sigmoid(z))
print("output-t:", tanh(z))
```

**Make reusable neuron**

```python
def neuron(x, weight, bias):
  z=np.dot(x,weights)+bias
  output = np.maximum(0,z)
  return output

x = np.array([2, 3, 1])
weights = np.array([0.5, 0.2, -0.4])
bias = 0.1

output = neuron(x, weights, bias)

print(output)
```

**Without np.dot() ----- By using loops**
