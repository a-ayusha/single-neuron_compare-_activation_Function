```python
import numpy as np
```

```python
# Activation functions
def relu(z):
    return np.maximum(0,z)

def sigmoid(z):
    return 1/(1+np.exp(-z))

# Forward propagation
def forward_propagation(X, W1, b1, W2, b2):
    # Hidden layer
    Z1= np.dot(X, W1)+b1
    A1 = relu(Z1)

    # Output layer
    Z2= np.dot(A1, W2)+b2
    A2= sigmoid(Z2)

    return(Z1, A1, Z2, A2)

# Example with 2 input features:

# X = np.array([[2.0,3.0]])
# W1= np.array([
#     [0.5, -1],
#     [1, 0.5]
# ])
# b1= np.array([[0,1]])

# # Output layer parameter:
# W2= np.array([
#     [0.7],
#     [-0.4]
# ])

# b2= np.array([[0.2]])

# Z1, A1, Z2, A2 = forward_propagation(X, W1, b1, W2,b2)

# print("Hidden weighted sums:\n", Z1)
# print("Hidden activations:\n", A1)
# print("Output weighted sum:\n", Z2)
# print("Final prediction:\n", A2)

```

```python
# Use two input features, three hidden neurons and one output neuron
X= np.array([[2,5]])
W1= np.array([
    [0.5, 1, 0.2],
    [0.2, 0.4, 2]
])
b1= np.array([[0,1,-0.5]])

W2= np.array([
    [0.2],
    [0.4],
    [0.5]
])
b2= np.array([[-0.2]])

Z1, A1, Z2, A2 = forward_propagation(X, W1, b1, W2,b2)

print("Hidden weighted sums:\n", Z1)
print("Hidden activations:\n", A1)
print("Output weighted sum:\n", Z2)
print("Final prediction:\n", A2)

```
