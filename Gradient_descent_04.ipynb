```python
import numpy as np
```

```python
w = 4
learning_rate = 0.1
epochs = 10

for epoch in range(epochs):
    loss = w**2
    dw= 2*w
    w = w-learning_rate * dw
    print(f"epoch{epoch+1}, Weight{w:.4f}, Loss{loss:.4f}")
```

```python
import matplotlib.pyplot as plt 
X = np.array([1,2,3,4], dtype=float)
y = np.array([3,5,7,9], dtype = float)

w= 0.0
b= 0.0
learning_rate = 0.01
epochs = 1000
#store loss values
loss_history = []

for epoch in range(epochs):
    #Forward propagation
    y_pred = w*X+b

    #Calculate MSE loss
    loss = np.mean((y_pred-y)**2)
    #Calculate gradients
    dw = (2/len(X))* np.sum(X*(y_pred-y))
    db = (2/len(X)) *np.sum(y_pred - y)
    #Update parameters
    w=w-learning_rate*dw
    b=b-learning_rate*db

    loss_history.append(loss)

    if (epoch+1)%100==0:
        print(f"epoch{epoch+1}, loss{loss:.6f}")

    #Final learned parameters:
print("\n Learned weight:", w)
print("Learned bias:", b)

#Final predictions:
print("Predictions:", w*X+b)
#plot loss
plt.plot(loss_history)
plt.xlabel("epochs")
plt.ylabel("MSE")
plt.title("Gradient descent - Loss reduction")
plt.show()
```
